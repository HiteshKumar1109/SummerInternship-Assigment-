# Q1.Write a Python function to find the maximum of three numbers.

```python
# MAximum number between three number
def maxi(demo):
    maxii=0
    for i in demo:
        if i>maxii:
            maxii=i
    return maxii
demo=[23,2,34]
maxi(demo)
demo2=[23,45,74,3,4,5]
maxi(demo2)
```

# Q4 . Write a Python function to calculate the factorial of a number (a non-negative integer). The function accepts the number as an argument

```python
# factorial of a number 
a=int(input("enter the number"))
def fact(a):
    if(a<=0):
        return 1
    while(a>0):
        return a*fact(a-1)
fact(a)    
```

# Q6. Write a Python function to check whether a number falls within a given range.


```python
def range(number,start,end):
    if start<=number<=end:
        print("number in the range")
    else:
        print("number out of the range")
number=int(input("enter the number"))
range(number,10,20)
range(number,15,20)
```

# Q8.  Write a Python function that takes a number as a parameter and checks whether the number is prime or not.


```python
number = int(input("Enter the number: "))

def prime(number, i=2):
    if number <= 1:
        return False
    
    if i == number:
        return True
    
    if number % i == 0:
        return False
    
    return prime(number, i + 1)

if prime(number):
    print("Number is prime")
else:
    print("Number is not prime")
```

# Q9.  Write a Python function that accepts a string and counts the number of upper and lower case letters. 


```python
str1=input("enter the string")
Upper_count=0
Lower_count=0
def count(str1):
    global Upper_count, Lower_count
    for i in str1:
        if ord(i)>=65 and ord(i)<=90:
            Upper_count+=1
        if ord(i)>=97 and ord(i)<=122:
            Lower_count+=1
count(str1)
print("upper letter are:",Upper_count)
print("lower letter are:",Lower_count)
```
