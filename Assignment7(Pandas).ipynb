```python
import pandas as pd
df3 = pd.DataFrame({
    'ID': [1, 2, 3, 4],
    'Name': ['Ram', 'Shyam', 'Aman', 'Riya']
})
df4 = pd.DataFrame({
    'ID': [1, 2],
    'Marks': [90, 85]
})
res3= df3.join(
    df4.set_index('ID'),
    on='ID',
    how='left'
)
print(res3)
```

# Q3 Create three DataFrames. Vertically concatenate two of them using pd.concat(), then merge the resulting DataFrame with the third DataFrame on a common key. T Understand join() vs. merge().
# Also Explain the primary differences between df.join() and pd.merge()

```python
# three data frame
dfA= pd.DataFrame({
    'ID': [1, 2, 3, 4],
    'Name': ['Ram', 'Shyam', 'Aman', 'Riya']
})
dfB = pd.DataFrame({
    'ID': [1,2,3,4],
    'Marks': [90, 85,90,89]
})
dfC=pd.DataFrame({
    'ID':[1,2,3,4],
    "class":["A","B","C","D"]
})
#concat the two dataFrame
resA = pd.concat(
    [dfA, dfB.drop(columns=["ID"])],
    axis=1
)
print(resA)

# merge that Frame with third dataframe
resB=pd.merge(resA,dfC,on="ID",how="left")
print("-------------------------")
print(resB)
```

difference between pd.join() && pd.merge()
merge()                     	             join()
Column-based	                             Index-based
Similar to SQL                               JOIN	Simpler for index joins
Uses on=	                                 Uses index automatically
Works on one or more columns	             Works mainly on index
More flexible	                             Faster/simple for indexes
