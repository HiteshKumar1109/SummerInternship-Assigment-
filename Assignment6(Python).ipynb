```python
#Create a Pandas DataFrame from List of Dicts
data4=[
    {"produt":"banana","price":900},
    {"product":"orange","price":700},
    {"product":"apple","price":1000}
]
df6=pd.DataFrame(data4)
print(df6)
print("-------------------------------------")
# finicial data frame
import pandas as pd

data4 = [
    {"product": "banana", "price": 900, "quantity": 10},
    {"product": "orange", "price": 700, "quantity": 15},
    {"product": "apple", "price": 1000, "quantity": 8}
]

df6 = pd.DataFrame(data4)

# new column
df6["total_price"] = df6["price"] * df6["quantity"]

# discount column
df6["discount"] = df6["price"] * 0.10

# final price after discount
df6["final_price"] = df6["price"] - df6["discount"]

print(df6)
```

# Q3 Data iteration
# 	Different ways to iterate over rows in Pandas Dataframe 
# 	Selecting rows in pandas DataFrame based on conditions 
# 	Select any row from a Dataframe using iloc[]
# 	Limited rows selection with given column
# 	Drop rows from the dataframe based on certain condition applied on a column 
# 	Insert row at given position in Pandas Dataframe 
# 	Create a list from rows in Pandas dataframe 

```python
#Different ways to iterate over rows in Pandas Dataframe
import pandas as pd
import numpy as np

# pandas work with the dictionary
data={
    "Name":['Hitesh','jatin','rahul','jay'],
    "age":[12,23,34,54],
    "salary":[2000,3000,4000,5000]
}
df=pd.DataFrame(data)
res=df.loc[:3]
print(res)
print("------------")
res2=df.iloc[0:2,0:1]
print(res2)
print("---------------")
print(df)
```

```python
# Selecting rows in pandas DataFrame based on conditions
import pandas as pd
import numpy as np

# pandas work with the dictionary
data={
    "Name":['Hitesh','jatin','rahul','jay'],
    "age":[12,23,34,54],
    "salary":[2000,3000,4000,5000]
}
df=pd.DataFrame(data)
print(df.loc[df['salary'] >= 3500, ['Name']])
```

```python
# Select any row from a Dataframe using iloc[]
list2=[
    ("Nike","Redtap","compus"),
    (8000,9000,900),
    ("smooth","good","average")
]
df2=pd.DataFrame(list2)
res=df2.iloc[0:1,0:3]
print(res)## select the one row
print("---------------")
# select the two ro from
res2=df2.iloc[0:2,0:3]
print(res2)
```

```python
#Limited rows selection with given column¶
import pandas as pd

data = {
    "Name": ["Hitesh", "jatin", "rahul", "jay"],
    "age": [12, 23, 34, 54],
    "salary": [2000, 3000, 4000, 5000]
}

df = pd.DataFrame(data)

print(df.loc[df['salary'] >= 3500, ['Name', 'salary']])
```

```python
# Drop rows from the dataframe based on certain condition applied on a column

data = {
    "Name": ["Hitesh", "Rahul", "Jay", "Aman"],
    "Marks": [90, 85, 70, 95],
    "City": ["Jaipur", "Delhi", "Mumbai", "Pune"]
}

df = pd.DataFrame(data)

print(df.loc[1:2, ['Name', 'Marks']])
res3=df.drop(1)
print(res3)
```

```python
# Insert row at given position in Pandas Dataframe
import pandas as pd

data = {
    "Name": ["Hitesh", "Rahul", "Jay"],
    "Marks": [90, 85, 70]
}

df = pd.DataFrame(data)

# new row
new_row = {
    "Name": "Aman",
    "Marks": 95
}

# split dataframe
upper = df.iloc[:1]
lower = df.iloc[1:]

# insert row
df_new = pd.concat([upper, pd.DataFrame([new_row]), lower]).reset_index(drop=True)

print(df_new)
```

```python
# Create a list from rows in Pandas dataframe¶
import pandas as pd
data7 = {
    "Name": ["Hitesh", "Rahul", "Jay"],
    "Marks": [90, 85, 70]
}

df7= pd.DataFrame(data7)
print(df7)
new_list=df7.values.tolist()
print(new_list)
```
