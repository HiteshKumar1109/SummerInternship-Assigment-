```python
#4) Write a Python program to calculate the total, highest value, and lowest value from a tuple of integers without using the built-in sum(), max(), and min() functions.
t=(10,20,30,40)
total = 0
highest = t[0]
lowest = t[0]
for i in t:
    total += i
    if i > highest:
        highest = i
    if i < lowest:
        lowest = i
print( total)
print(highest)
print( lowest)
```
