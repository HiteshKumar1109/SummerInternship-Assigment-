```python
from pyspark.sql import SparkSession
import pandas as pd
import os
```

```python
spark=SparkSession.builder.appName("spark_app").getOrCreate()
```

```python
data = """product_id,product_name,category,sales
101,Laptop,Electronics,150000
102,Mobile,Electronics,95000
103,TV,Electronics,120000
104,Chair,Furniture,30000
105,Table,Furniture,45000
106,Sofa,Furniture,80000
107,Headphones,Electronics,25000
108,Bed,Furniture,90000
"""
```

```python
os.makedirs("/workspace/Soucre_data/",exist_ok=True)
file_path=os.path.join("/workspace/Soucre_data/","products.csv")
with open(file_path,"w")as file:
    file.write(data)
print("fiel saved into the",file_path)
```

```python
# read the file from the product data
read_data=spark.read.csv("Soucre_data/products.csv",header=True,inferSchema=True)
```

```python
read_data.createOrReplaceTempView("products")
```

```python
read_data.show()
```

```python
from pyspark.sql.functions import col
```

```python
sort_data=read_data.sort(col("sales").desc())
```

```python
sort_data.show()
```

```python
sort_data.write \
    .mode("overwrite") \
    .option("header", True) \
    .csv("/workspace/Assinment17_output/Sort_data")
```

```python
top3_product=sort_data.orderBy(col("sales").desc()).limit(3)
```

```python
top3_product.show()
```

```python
top3_product.write \
     .mode("overwrite") \
     .option("header",True) \
     .csv("/workspace/Assinment17_output/top_product")
```

```python
# now take those product  their sales have more then 80,000
filtered_data=read_data.filter(col("sales")>80000)
```

```python
filtered_data.show()
```

```python
filtered_data.write \
     .mode("overwrite") \
     .option("header",True) \
     .csv("/workspace/Assinment17_output/filteredData")
```
