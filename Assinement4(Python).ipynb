# Q1.Create a CSV file for address book, CSV file should have column for name, address, mobile, email. Insert 2-3 dummy data entered by user.

```python
# create a csv file and store the 2 to 4 line of row in the file
import csv
data=[
    ["Name","Address","Mobile","Email"],
    ["Rahul","Ajmer",6389657896,"Rahul@gmail.com"],
    ["Jatin","Hanumangarh",8978986545,"Jatin@gmail.com"],
    ["Ajay","Bihar",9654696867,"Ajay@gmail.com"],
    ["Ravi","Pilibanga",8967989678,"Ravi@gmail.com"]
]
with open("demo.csv","w") as file:
    writer=csv.writer(file)
    for i in data:
        writer.writerow(i)
```

```python
# read the file
import pandas as pd
with open("demo.csv","r")as file:
    df=pd.read_csv(file)
print(df)
    
```
