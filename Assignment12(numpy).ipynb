# 1) Replace Nan with 0 and Interchange 3 rows and 3 columns of 2D array [[6, -8, 73, -110], [np.nan, -8, 0, 94]]

```python
import numpy as np
arr1=np.array([[6, -8, 73, -110], [np.nan, -8, 0, 94]])
print(arr1)
res=np.nan_to_num(arr1,nan=0)
print("-------------------------")
print(res)
```

```python
# replace the third row with third column
res[:,[2,3]]=res[:,[3,2]]
print(res)
```

# 3) Replace NaN values with average of columns


```python
arr2=np.array([[6, -8, 73, -110], [np.nan, -8, 0, 94]])
print(arr2)
```

```python
mean=np.nanmean(arr2)
arr2[np.isnan(arr2)]=mean
print(arr2)
```

```python
arr3= np.array([[6, -8, 73, -110],
                [5, -8, 0, 94]])
arr3[arr3<0]=0
print(arr3)
```

# 5) Study the following program import numpy as np # create a numpy 1d-arrays arr1 = np.array([3, 4]) arr2 = np.array([1, 0]) # find average of NumPy arrays avg = (arr1 + arr2) / 2 print("Average of NumPy arrays:\n", avg) -> Calculate average mean median mode values of two NumPy 2d-arrays 

```python
demo1=np.array([3,4])
demo2=np.array([1,0])
avg=(demo1+demo2)/2
print("Average Value of NUmpy Arrays:\n:",avg)
```

```python
from scipy import stats
print("Average Array:")
print(avg)

print("\nMean of Average Array:")
print(np.mean(avg))

print("\nMedian of Average Array:")
print(np.median(avg))

print("\nMode of Average Array:")
print(stats.mode(avg, axis=None, keepdims=False))
```

#  6) Solve the following equation using linalg() and inverse Matrix method x - 2y + 3z = 9 -x + 3y - z = -6 2x - 5y + 5z = 17 7) Using "Customizing Matplotlib Visualizations" discussed in Numpy session4 notes plot graph to compare your at least 2 semester result

```python
A = np.array([
    [1, -2, 3],
    [-1, 3, -1],
    [2, -5, 5]
])

B = np.array([9, -6, 17])

X = np.linalg.solve(A, B)

print("x =", X[0])
print("y =", X[1])
print("z =", X[2])
```

```python
import matplotlib.pyplot as plt
variables = ["x", "y", "z"]
values = X

plt.figure(figsize=(6, 4))
plt.bar(variables, values)

plt.title("Solution of Linear Equations")
plt.xlabel("Variables")
plt.ylabel("Values")
plt.grid(True)

plt.show()
```

```python
#line chart
plt.figure(figsize=(6, 4))
plt.plot(variables, values)

plt.title("Solution of Linear Equations")
plt.xlabel("Variables")
plt.ylabel("Values")
plt.grid(True)

plt.show()
```
