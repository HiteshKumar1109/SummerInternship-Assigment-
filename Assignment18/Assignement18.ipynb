# Create a PySpark application that generates a DataFrame containing 5 million records using spark.range(). Display the initial number of partitions, then increase the partitions to 12 using repartition(). After that, reduce the partitions to 3 using coalesce()

```python
from pyspark.sql import SparkSession
import pandas as pd
```

```python
spark=SparkSession.builder.appName("Demo_app").getOrCreate()
```

```python
df=spark.range(0,5000000)
```

```python
print("Intitally number of partitions",df.rdd.getNumPartitions())
```

```python
df.write.mode("overwrite").csv("/workspace/Assign_output/Original_data")
```

```python
# Do 12 partition of the data
df_repartitions=df.repartition(12)
```

```python
print("partitons after repartitions",df_repartitions.rdd.getNumPartitions())
```

```python
# store the repartitioned data
df_repartitions.write.mode("overwrite").csv("/workspace/Assign_output/Repartition_data")
```

#### perform coalesce()

```python
df_coalesce=df_repartitions.coalesce(3)
```

```python
print("The partititons after the coalesce in the data",df_coalesce.rdd.getNumPartitions())
```

```python
df_coalesce.write.mode("overwrite").csv("/workspace/Assign_output/Coalesce_data")
```
