# Q2 Try building today's game yourself


```python
import random

choices = ["rock", "paper", "scissor"]

user_win = 0
computer_win = 0
draw = 0

while True:
    user = input("Enter rock, paper, scissor or quit: ").lower()

    if user == "quit":
        print("Game Over")
        print("Your Wins:", user_win)
        print("Computer Wins:", computer_win)
        print("Draws:", draw)
        break

    if user not in choices:
        print("Invalid Choice")
        continue

    computer = random.choice(choices)

    print("Computer:", computer)

    if user == computer:
        print("Draw")
        draw += 1

    elif (
        (user == "rock" and computer == "scissor") or
        (user == "paper" and computer == "rock") or
        (user == "scissor" and computer == "paper")
    ):
        print("You Win")
        user_win += 1

    else:
        print("Computer Wins")
        computer_win += 1

    print("Score -> You:", user_win, "Computer:", computer_win, "Draw:", draw)
```
