```python
#Create a 3×3 NumPy array of all 9
res2=np.full((3,3),9)
print(res2)
```

```python
#Create a 1D array of 10 evenly spaced values between 25 and 125
res3=np.arange(25,125,10)
print(res3)
```

```python
# Convert a Python list into a NumPy array
list1=[1,2,3,4,5,6]
res4=np.array(list1)
print(res4)
print(res4.ndim)
print(res4.shape)
print(res4.data)
```

```python
#Reverse a 1D NumPy array 
arr2=np.array([23,34,45,55,67,78])
print(np.flip(arr2))
```

```python
#Create a 4×4×3 array and extract value at its second set, first row and last column
res5=np.arange(48).reshape(4,4,3)
print(res5)
```

```python
#Create a 4×4 and Extract Odd Rows and Even Columns
res6=np.arange(16).reshape(4,4)
print(res6)
```

```python
print(res6[1::2,::2])
```

```python
#Slice the first two rows and first two columns of econd set from a 4×4×3 array
res5=np.arange(48).reshape(4,4,3)
print(res5[:,:2,:2])
```

```python
# Replace all odd numbers in a NumPy array with -1 by itterating using for loop [[23, 56, 78, 93], [71, 82,13, 24]]
arr4=np.array([[23, 56, 78, 93], [71, 82,13, 24]])
print(arr4)
```

```python
for i in range(arr4.shape[0]):
    for j in range(arr4.shape[1]):
        if arr4[i][j] % 2 != 0:
            arr4[i][j] = -1
print(arr4)
```

```python
# Get the indices of non-zero elements in an array [1, 0, 2, 0, 3, 0, 4] 
arr6=np.array([1, 0, 2, 0, 3, 0, 4])
print(arr6)
print(np.nonzero(arr6))
```

```python
for idx in np.ndindex(arr6.shape):
    if arr6[idx]!=0:
        print(idx)
```

```python
# Write a code to compute the dot product of two NumPy arrays arr1 = [15, 20, 25] arr2 = [10,40,37]
demo3=np.array([15, 20, 25])
demo4=np.array([10,40,37])
result=np.dot(demo3,demo4)
print(result)
```
