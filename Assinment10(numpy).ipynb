# Q1  Combining a one and a two-dimensional NumPy Array

```python
# 1-d array
import numpy as np
arr1=np.array([1,2,3,4,5,6])
arr2=np.array(([34,56,78],[23,45,67]))
arr1.ndim
print(arr1)
arr2.ndim
print(arr2)
```

```python
# concate the numpy array in the 
arr_1=arr1.reshape(2,3)
arr3 = np.concatenate((arr_1, arr2),axis=0)
print(arr3)
```

```python
# concatenate the array horizontally
arr4=np.concatenate((arr_1,arr2), axis=1)
print(arr4)
```

```python
# reverse the numpy 2-d arrray
exp2=np.array([[1,2,3,4],[6,7,8,9]])
print(exp2)
```

```python
# us the fliud() function to flip the array
exp3=np.flipud(exp2)
exp4=np.fliplr(exp2)
exp5=np.rot90(exp2)
print(exp3)
print("---------------")
print(exp4)
print("---------------")
print(exp5)
```
